```python
import pandas as pd
import numpy as np
import re
import string
```

```python
data = pd.read_excel(r'/content/TELIKO_SERRES.xlsx')
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df['type'] = df['type'].str.split(',').str[0]
```

```python
df['price'] = df['price'].str.lstrip('€')
```

```python
df['area'] = df['area'].str.extract(r'(\d+)')
```

```python
df['price'] = df['price'].str.replace('.','',regex=False)
```

```python
df['location'] = df['location'].str.split(',').str[0]
```

```python
df['location'].unique()
```

```python
df.head(5)
```

```python
df['bedrooms'] = df['bedrooms'].str.extract(r'(\d+)')
df
```

```python
df['bedrooms'].isnull().sum()
```

```python
df['floor']
```

```python
df['floor'] = df['floor'].replace({'Ισόγειο': 0, 'Ημιόροφος': 0.5})
df['floor'] = df['floor'].replace({'1ος': 1, '2ος': 2, '3ος': 3, '4ος': 4})
df
```

```python
df['heating'] = df['heating'].apply(lambda x: x.split('(')[-1].split(')')[0] if isinstance(x, str) else x)
```

```python
df['energy_class'] = df['energy_class'].replace({'Εξαιρείται από την υποχρεωτική έκδοση ΠΕΑ':np.NaN,'Το ΠΕΑ είναι υπό έκδοση':np.NaN})
```

```python
df['energy_class'] = df['energy_class'].str[0]
```
